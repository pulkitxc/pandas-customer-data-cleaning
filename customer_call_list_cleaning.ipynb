```python
import pandas as pd
```

```python
df = pd.read_excel(r"C:\Users\Pallav\Downloads\Customer Call List.xlsx")
df
```

```python
df = df.drop_duplicates()
df
```

```python
df = df.drop(columns="Not_Useful_Column", errors='ignore')
df
```

```python
df["Last_Name"] = df["Last_Name"].str.strip("123._/")
df
```

```python
df["Phone_Number"] = df["Phone_Number"].str.replace(r"\D","",regex = True)
df["Phone_Number"]
```

```python
df["Phone_Number"].unique()
```

```python
df.dropna(subset=["Phone_Number"])
df
df["Phone_Number"] = df["Phone_Number"].replace("", pd.NA)
df
```

```python
df["Phone_Number"].unique()
```

```python
df["Phone_Number"] = df["Phone_Number"].apply(lambda x : str(x)) 
df
```

```python
df["Phone_Number"].head()
```

```python
df["Phone_Number"] = df["Phone_Number"].replace(["nan", "<NA>"], pd.NA)
df
```

```python
df["Phone_Number"].unique()
```

```python
mask = df["Phone_Number"].notna()

df.loc[mask, "Phone_Number"] = (
    df.loc[mask, "Phone_Number"].str[:3]
    + "-"
    + df.loc[mask, "Phone_Number"].str[3:6]
    + "-"
    + df.loc[mask, "Phone_Number"].str[6:]
)
```

```python
df["Phone_Number"] = df["Phone_Number"].apply(lambda x : str(x))
df
```

```python
df["Phone_Number"] = df["Phone_Number"].str.replace("<NA>", '')
df
```

```python
df[['Street_Address','State','Zip_Code']] = df["Address"].str.split(",", n=2, expand=True)
df
```

```python
df = df.fillna('')
df
```

```python
df['Paying Customer'] = df['Paying Customer'].replace('Yes', 'Y')
df['Paying Customer'] = df['Paying Customer'].replace('No', 'N')
df['Do_Not_Contact'] = df['Do_Not_Contact'].replace('Yes', 'Y')
df['Do_Not_Contact'] = df['Do_Not_Contact'].replace('N0', 'N')
df['Do_Not_Contact'] = df['Do_Not_Contact'].replace('No', 'N')
df['Paying Customer'] = df['Paying Customer'].replace('Y', 'Yes')
df['Paying Customer'] = df['Paying Customer'].replace('N', 'No')
df['Do_Not_Contact'] = df['Do_Not_Contact'].replace('Y', 'Yes')
df['Do_Not_Contact'] = df['Do_Not_Contact'].replace('N', 'No')
df
```

```python
df['Paying Customer'] = df['Paying Customer'].replace('N/a', '')
df['Address'] = df['Address'].replace('N/a', '')
df
```

```python
for x in df.index:
    if df.loc[x,"Do_Not_Contact"] == 'Y' :
        df.drop(x,inplace = True)
df
```

```python
for x in df.index:
    if df.loc[x,"Phone_Number"] == '' :
        df.drop(x,inplace = True)
df
```

```python
df = df.reset_index(drop=True)
df
```

```python
df.to_csv("Cleaned_Customer_Call_List.csv", index=False)
```
